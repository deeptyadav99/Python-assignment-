# 7.what is .pyc file (byte code) when it is created?

When Python runs your code, it converts your .py file into bytecode. This bytecode can be stored in a .pyc file so Python doesn't have to compile the same code again when it is imported later.

# 8. what is pvm?

* PVM stands for Python Virtual Machine.
* PVM is the part of Python that runs the bytecode.   

# 9. how python internally works?

* Python internally works in 3 main steps.
    1.Write Python code - You write your code in a .py file.
    2.Compilation - Python converts your source code into bytecode.
    3.Execution - The PVM (Python Virtual Machine) executes the bytecode and produces the output.
